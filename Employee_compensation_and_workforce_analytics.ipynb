```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("hr_workforce_dataset.csv")
print("Dataset loaded!")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["commission_pct"] = df["commission_pct"].fillna(0)
```

```python
df["department_name"] = df["department_name"].fillna("Unassigned")
df["city"] = df["city"].fillna("Unassigned")
df["state_province"] = df["state_province"].fillna("Unassigned")
df["country_name"] = df["country_name"].fillna("Unassigned")
df["region_name"] = df["region_name"].fillna("Unassigned")
```

```python
df["manager_name"] = df["manager_name"].fillna("No Manager")
```

```python
df.isnull().sum()
```

```python

df["manager_id"] = df["manager_id"].fillna(0)
df["department_id"] = df["department_id"].fillna(0)

df.isnull().sum()
```

```python
df["hire_date"] = pd.to_datetime(df["hire_date"])
```

```python
df["hire_date"].head()
```

```python
df.dtypes
```

```python
df["hire_year"] = df["hire_date"].dt.year
```

```python
df[["employee_name", "hire_date", "hire_year"]].head()
```

```python
df["salary_band"] = pd.qcut(
    df["salary"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
df[["employee_name", "salary", "salary_band"]].head()
```

```python
df["salary_band"].value_counts()
```

```python
df["has_commission"] = (
    df["commission_pct"] > 0
).map({
    True: "Yes",
    False: "No"
})
```

```python
df["has_commission"].value_counts()
```

```python
df["pay_range_position"] = (
    (df["salary"] - df["min_salary"])
    / (df["max_salary"] - df["min_salary"])
).round(2)
```

```python
df[
    [
        "employee_name",
        "job_title",
        "salary",
        "min_salary",
        "max_salary",
        "pay_range_position"
    ]
].head()
```

```python
df["pay_level"] = pd.cut(
    df["pay_range_position"],
    bins=[-0.01, 0.5, 1.0],
    labels=["Below Midpoint", "Above Midpoint"]
)
```

```python
df["pay_level"].value_counts()
```

```python
print(
    "Number of different job IDs:",
    df["job_id"].nunique()
)
```

```python
print(
    "Number of different job titles:",
    df["job_title"].nunique()
)
```

```python
df = df.drop(
    columns=[
        "job_id",
        "email",
        "phone_number"
    ]
)
```

```python
print("Rows and columns after cleaning:")
print(df.shape)
```

```python
df.head()
```

```python
employees_by_department = (
    df["department_name"]
    .value_counts()
)
print("Employees by dept:")
print(employees_by_department)
```

```python
average_salary_by_department = (
    df.groupby("department_name")["salary"]
    .mean()
    .sort_values(ascending=False)
)
print("average_salary_by_department:")
print(average_salary_by_department)
```

```python
employees_by_region = (
    df["region_name"]
    .value_counts()
)
print("employees_by_region:")
print(employees_by_region)
```

```python
commission_count = (
    df["has_commission"]
    .value_counts()
)
print("commission_count:")
print(commission_count)
```

```python
total_payroll = df["salary"].sum()
print("Total payroll:", total_payroll)
```

```python
average_salary = df["salary"].mean()
print("Average salary:", round(average_salary, 2))
```

```python
average_salary = df["salary"].mean()

employees_above_average = df[
    df["salary"] > average_salary
][
    [
        "employee_name",
        "job_title",
        "salary"
    ]
].sort_values(
    "salary",
    ascending=False
)
print("employees_above_average salary:")
print(employees_above_average)
```

```python
salary_by_region = (
    df.groupby("region_name")["salary"]
    .agg(
        ["count", "mean", "sum"]
    )
    .sort_values(
        "mean",
        ascending=False
    )
)
print("salary by region:")
print(salary_by_region)
```

```python
salary_by_commission = (
    df.groupby("has_commission")["salary"]
    .agg(
        ["count", "mean"]
    )
)
print("salary by commission status:")
print(salary_by_commission)
```

```python
payroll_by_band = (
    df.groupby("salary_band", observed=False)["salary"]
    .agg(
        ["count", "sum"]
    )
    .sort_values(
        "sum",
        ascending=False
    )
)
print("payroll by salary band:")
print(payroll_by_band)
```

```python
employees_above_midpoint = df[
    df["pay_level"] == "Above Midpoint"
][
    [
        "employee_name",
        "job_title",
        "salary",
        "pay_range_position"
    ]
].sort_values(
    "pay_range_position",
    ascending=False
)

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 2000)
pd.set_option("display.max_colwidth", 30)

print("employees above job midpoint:")
print(employees_above_midpoint)
```

```python
manager_team_size = (
    df[
        df["manager_name"] != "No Manager"
    ]
    .groupby("manager_name")["employee_id"]
    .count()
    .sort_values(
        ascending=False
    )
)
print("managers with largest team size:")
print(manager_team_size.head(5))
```

```python
employees_with_previous_jobs = df[
    df["previous_jobs_count"] > 0
][
    [
        "employee_name",
        "job_title",
        "previous_jobs_count"
    ]
].sort_values(
    "previous_jobs_count",
    ascending=False
)
print("employees with previous job history:")
print(employees_with_previous_jobs)
```

```python
payroll_by_country = (
    df.groupby("country_name")["salary"]
    .sum()
    .sort_values(
        ascending=False
    )
)
print("payroll by country:")
print(payroll_by_country)
```

```python
top_5_salary = (
    df.sort_values(
        "salary",
        ascending=False
    )
    [
        [
            "employee_name",
            "job_title",
            "salary"
        ]
    ]
    .head(5)
)
print("first 5 highest paid employees:")
print(top_5_salary)
```

```python
plt.figure(figsize=(7, 4))

sns.histplot(
    df["salary"],
    bins=15
)

plt.title("Distribution of Salaries")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(
    x=df["salary"]
)

plt.title("Salary Box Plot")
plt.xlabel("Salary")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

(
    df.groupby("department_name")["salary"]
    .mean()
    .sort_values(ascending=False)
    .plot(kind="bar")
)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")

plt.xticks(
    rotation=60,
    ha="right"
)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(5, 5))

df["region_name"].value_counts().plot(
    kind="pie",
    autopct="%1.0f%%"
)

plt.title("Employees by Region")
plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

(
    df["hire_year"]
    .value_counts()
    .sort_index()
    .plot(
        kind="line",
        marker="o"
    )
)

plt.title("Number of Hires by Year")
plt.xlabel("Year")
plt.ylabel("Employees Hired")

plt.show()
```

```python
correlation_data = df[
    [
        "salary",
        "commission_pct",
        "hire_year",
        "previous_jobs_count",
        "pay_range_position"
    ]
]

correlation_matrix = correlation_data.corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
from sqlalchemy import create_engine
```

```python
connection_string = "mysql+pymysql://"
connection_string = connection_string + user
connection_string = connection_string + ":"
connection_string = connection_string + password
connection_string = connection_string + "@"
connection_string = connection_string + host
connection_string = connection_string + "/"
connection_string = connection_string + database
```

```python
engine = create_engine(connection_string)
print("connected to database successfully !")
```

```python
table_name = "hr_workforce"
```

```python
df.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)
print("Data uploaded to MySQL successfully.")
```

```python
sql_data = pd.read_sql(
    "SELECT * FROM hr_workforce",
    engine
)
print("reading data from mysql:")
print(sql_data.head())
```
